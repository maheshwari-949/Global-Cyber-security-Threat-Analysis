```python
import pandas as pd
df = pd.read_csv(r"C:\Users\SHIVA\DESKTOP\cybersecurity data\global cs threats\Global_Cybersecurity_Threats_2015-2024.csv")
df.head()
```

```python
# check missing values
df.isnull().sum()
```

```python
# clean column  names and remove duplicates
df.columns = df.columns.str.strip()
df = df.drop_duplicates()
```

```python
# Summary statistics using .describe()
df. describe(include='all')
```

```python
# Renaming columns according to snake casing for better readability and documentation
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.info()
df.isnull()
```

```python
df.info()
```

```python
# Rename columns
df = df.rename(columns={
    'security_vulnerability_type': 'vulnerability_type',
    'defense_mechanism_used': 'defense_mechanism',
    'financial_loss_(in_million_$)': 'financial_loss_m',
    'incident_resolution_time_(in_hours)': 'resolution_time_h',
    'number_of_affected_users': 'affected_users',
    
})

print(df.columns.tolist())
```

```python
# checking what type of attacks are there and how many times it occurred
print(df['attack_type'].value_counts())
print(df['attack_type'].unique())
```

```python
#  Creating new column with simple names for attacks for dashboard - easy for anyone to understand
attack_map_pro = {
    'Phishing': 'Email Fraud',
    'Ransomware': 'Data Lock',
    'DDoS': 'Site Closed',
    'SQL Injection': 'Database Attack',
    'Malware': 'virus Attack',
    'Man-in-the-Middle': 'Message Stolen'
}

df['attack_category'] = df['attack_type'].map(attack_map_pro)

print(df['attack_category'].value_counts())
```

```python
# Creating new column with simple names
attack_map_pro = {
    'Zero-day': 'Unknown Bug',
    'Social Engineering': 'people Tricked',
    'Unpatched Software': 'System Not Updated',
    'Weak Passwords': 'Weak Password'
}

df['vulnerability_category'] = df['vulnerability_type'].map(attack_map_pro)

print(df['vulnerability_category'].value_counts())
```

```python
print(df.columns.tolist())
    
```

```python
def get_severity(x):
    if x > 50:
        return 'high'
    elif x >= 20:
        return 'moderate'
    else:
        return 'low'

df['loss_severity'] = df['financial_loss_m'].apply(get_severity)
```

```python
df['resolution_speed'] = df['resolution_time_h'].apply(lambda x: 'Fast' if x < 24 else 'Slow')
```
